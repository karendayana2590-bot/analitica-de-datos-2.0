## Un DataFrame organiza registros en filas y columnas

Cada columna puede contener un tipo de dato diferente y cada fila reúne un registro.

Propósito: Construir una tabla pequeña y explicar de dónde sale cada fila.

<p>Un <strong>DataFrame</strong> es una estructura tabular de dos dimensiones. Los nombres de columna describen las variables y el índice etiqueta las filas. Al seleccionar una sola columna, obtenemos normalmente una Series.</p><p>Vamos a crear una tabla con turno y cantidad. El texto A y B identifica categorías. Los números 100 y 90 representan unidades. En esta construcción, los primeros elementos de las listas forman la primera fila.</p>

### Reunir dos variables para consultar registros completos.

Preparación que necesita el ejemplo:


```python
import pandas as pd
```

```python
mini = pd.DataFrame({
    "turno": ["A", "B"],
    "cantidad": [100, 90]
})
mini
```

**Qué hace cada línea**

1. pd.DataFrame inicia la creación de una tabla. La llave abre un diccionario de nombres y valores.

2. "turno" es el nombre de la primera columna. Los dos puntos lo asocian con la lista ["A", "B"]. La coma separa esta columna de la siguiente.

3. "cantidad" se asocia con los números 100 y 90. Las dos listas tienen dos elementos.

4. La llave cierra el diccionario y el paréntesis cierra la construcción de DataFrame.

5. mini muestra el objeto guardado con ese nombre.

**Resultado esperado:** Dos filas y dos columnas: A con 100, B con 90.

**Interpretación:** La cantidad y el turno quedan en el mismo registro según su posición en las listas. Los nombres de columna describen qué significa cada valor.

**Error y corrección:** ValueError aparece si las listas tienen distintas longitudes. Antes de corregir, comprueba qué registro carece de un valor; no inventes una cantidad para igualarlas.

**Decisión:** Exigir que cada registro tenga los campos necesarios antes de incorporarlo a una tabla.
